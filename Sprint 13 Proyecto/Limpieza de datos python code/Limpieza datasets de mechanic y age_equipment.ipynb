```python
import pandas as pd
```

```python
df_age_equipment = pd.read_csv('../Dataset/age_equipment.csv')
df_mechanic = pd.read_csv('../Dataset/mechanic.csv')
```

## Revisión de datos y resolución de problemas de tipificación

#### Dataframe **df_age_equipment** con datos de antigüedad de maquinas de costura.

```python
print('Número de registros en archivo age_equipment.csv: ', df_age_equipment.shape[0])

if 'Unnamed: 0' in df_age_equipment.columns:
    df_age_equipment.drop(columns='Unnamed: 0', inplace=True)
df_age_equipment.head(5)
```

#### Revisión de valores nulos.

```python
df_age_equipment.info()
```

```python
# Valores nulos por variable
nulos_age_equip = df_age_equipment.isnull().sum()

# Calcular el porcentaje de nulos
nulos_age_equip_pers = (nulos_age_equip / len(df_age_equipment)) * 100

null_info_age_equip = pd.DataFrame({
    'Variable': nulos_age_equip.index,
    'Nulos': nulos_age_equip.values,
    'Porcentaje de Nulos': nulos_age_equip_pers.values
})

null_info_age_equip
```

```python
df_age_equipment[df_age_equipment['Age'].isnull()].head(10)
```

Eliminamos las filas con nulos.

```python
df_age_equipment.dropna(axis=0, inplace=True)
```

```python
df_age_equipment.info()
```

Cambiamos el tipo de ambas columas por int.

```python
df_age_equipment['Age'] = df_age_equipment['Age'].astype(int)
df_age_equipment['Equipment'] = df_age_equipment['Equipment'].astype('int')
```

Cambiamos primeras letras de los nombres de columnas por minúsculas.

```python
df_age_equipment.columns = df_age_equipment.columns.str.lower()
```

```python
df_age_equipment.head(5)
```

#### **Número de maquinas** con información de **antigúedad.**

```python
print('Número de maquinas con información de antigúedad:', df_age_equipment.shape[0])
```

#### Dataframe **df_mechanic** con nombres de mecánicos y sus experiencia laboral en la fabrica.

```python
if 'Unnamed: 0' in df_mechanic.columns:
    df_mechanic.drop(columns='Unnamed: 0', inplace=True)

df_mechanic.sort_values(by='mechanic').head(10)
```

```python
df_mechanic.info()
```

#### Revisión de valores nulos.

```python
# Valores nulos por variable
nulos_mechanic = df_mechanic.isnull().sum()

# Calcular el porcentaje de nulos
nulos_mechanic = (nulos_mechanic / len(df_mechanic)) * 100

null_info_mechanic = pd.DataFrame({
    'Variable': nulos_mechanic.index,
    'Nulos': nulos_mechanic.values,
    'Porcentaje de Nulos': nulos_mechanic.values
})

null_info_mechanic
```

#### Cambiamos el tipo de variable **mechanic** por category

```python
df_mechanic['mechanic'] = df_mechanic['mechanic'].astype('category')
```

#### **Número de mecánocos** con información de **experiencia.**

```python
print('Número de mecánicos con información de experiencia:', df_mechanic.shape[0])
```

#### Guardamos datasets limpiados

```python
df_age_equipment.to_parquet('../Dataset limpiado/age_equipment_limp.parquet')
df_mechanic.to_parquet('../Dataset limpiado/mechanic_limp.parquet')
```
