## Revisión de datos y resolución de problemas de tipificación

#### Dataframe **df_orden** con datos detallados sobre el historal de fallas de maquinas de costura.

#### Lista de variables con el número de valores únicos y ejemplos de valores.

```python
variables = pd.DataFrame(columns=['Variable', 'Número de valores unicos', 'Valores'])
for i, var in enumerate(df_orden.columns):
    variables.loc[i] = [var, df_orden[var].nunique(), df_orden[var].unique().tolist()]

variables
```

#### Revisión de valores nulos.

Se ve que dataframe no tiene valores nulos.

```python
# Valores nulos por variable
nulos_df_orden = df_orden.isnull().sum()

# Calcular el porcentaje de nulos
nulos_df_orden = (nulos_df_orden / len(df_orden)) * 100

null_info = pd.DataFrame({
    'Variable': nulos_df_orden.index,
    'Nulos': nulos_df_orden.values,
    'Porcentaje de Nulos': nulos_df_orden.values
})

null_info
```

#### Variables **problem_code, failure_code, cause_code, action_code** duplican la informacion de las columnas **problem_desc, failure_desc, cause_desc, action_desc** respectivamente. 
#### Para facilitar la interpretación quedamos con columnas descriptivas.

```python
df_orden.drop(columns=['problem_code', 'failure_code', 'cause_code', 'action_code'], inplace=True)
```

#### Python ha tipificado las variables **problem_desc, failure_desc, cause_desc, action_desc** como object.

#### Cambiamos el tipo de variables por category.

```python
df_orden[['problem_desc', 'failure_desc', 'cause_desc', 'action_desc']] = \
    df_orden[['problem_desc', 'failure_desc', 'cause_desc', 'action_desc']].astype('category')
```

#### Variables **operation, wo_type, base_model, mechanic** son también categóricas.

```python
df_orden[['operation', 'wo_type', 'base_model', 'mechanic']] = \
    df_orden[['operation', 'wo_type', 'base_model', 'mechanic']].astype('category')
```

#### Variable **parts_cost**.

Eliminamos símbolos de **$** , paréntesis de apertura y de cirre y comas y convertimos a float.

```python
', '.join(df_orden['parts_cost'].head(20).astype(str))
```

```python
df_orden['parts_cost'] = df_orden['parts_cost'].str.replace('$', '').str.replace(',', '').str.replace('(', '').str.replace(')', '').astype(float)
```

```python
', '.join(df_orden['parts_cost'].head(20).astype(str))
```

#### Variable **date**.

```python
df_orden['date'].info()
```

```python
', '.join(df_orden['date'].head(20).astype(str))
```

Convertimos los valores de la **date** a datetime.

```python
df_orden['date'] = pd.to_datetime(df_orden['date'])
df_orden['date'].info()
```

```python
', '.join(df_orden['date'].head(20).astype(str))
```

#### Variable **downtime_gross**.

Observamos que hay valores negativos de la **downtime_gross**.

```python
df_orden['downtime_gross'][df_orden['downtime_gross'] < 0]
```

Eliminamos registros con valores negativas.

```python
df_orden = df_orden.drop(index = df_orden['downtime_gross'][df_orden['downtime_gross'] < 0].index)
```

#### Variable **waiting_hours**.

Observamos que hay valores negativos de la **waiting_hours**.

```python
df_orden['waiting_hours'][df_orden['waiting_hours'] < 0]
```

Eliminamos registros con valores negativas.

```python
df_orden = df_orden.drop(index = df_orden['waiting_hours'][df_orden['waiting_hours'] < 0].index)
```

#### Variable **mechanic**

```python
filtro1 = df_orden['mechanic'] == 'GALEAS PEREZ EDWIN ROGELIO'
filtro2 = df_orden['mechanic'] == 'Edwin Rogelio Galeas Perez'
```

```python
df_orden['mechanic'][filtro1 | filtro2 ]
```

Cambiamos el nombre de mecánico porque está escrito en dos formas.

```python
df_orden['mechanic'][filtro1] = 'Edwin Rogelio Galeas Perez'
```

#### Agregamos la variable **days_between_failures**.

```python
df_orden = df_orden.sort_values(["equipment","date"])
```

```python
df_orden['days_between_failures'] = df_orden.groupby("equipment")["date"].diff().dt.days
```

```python
df_orden['days_between_failures'].isnull().sum()
```

```python
df_orden = df_orden.dropna(subset=['days_between_failures']).reset_index()

if 'index' in df_orden.columns:
    df_orden.drop(columns='index', inplace=True)

# df_orden['days_between_failures'] = df_orden['days_between_failures'].fillna( (df_orden['date'] - pd.Timestamp('2023-01-01') ).dt.days )
# df_orden['days_between_failures'] = df_orden['days_between_failures'].astype(int)
```
