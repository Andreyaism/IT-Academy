```python
def getEdad(fecha_nacim):
    hoy = date.today()
    return hoy.year - fecha_nacim.year - ( (hoy.month, hoy.day) < (fecha_nacim.month, fecha_nacim.day) )
```

```python
def getTypeVar(serie, umbral_categorica):

    # Devuelve el tipo de variable:
    # - 'numerica_continua'
    # - 'numerica_discreta'
    # - 'categorica'
    # - 'temporal'
   
    if pd.api.types.is_datetime64_any_dtype(serie):
        return "temporal"
    
    if pd.api.types.is_numeric_dtype(serie):
        unique_ratio = serie.nunique() / len(serie)
        if unique_ratio < umbral_categorica:
            return "numerica_discreta"
        else:
            return "numerica_continua"
        
    return "categorica"
```

```python
def exportarGraficos(my_df, carpeta):

    # Creamos carpeta (si no existe)
    if not os.path.exists(carpeta):
        os.makedirs(carpeta)
        
    # Iteramos por columnas de dataframe para datectar el tipo de cada colunna    
    for col in my_df:
        plt.figure(figsize=(6, 4))
        tipo_var = getTypeVar(my_df[col], 0.05)
        if tipo_var in ["numerica_discreta", "categorica"]:
            my_df[col].value_counts().plot.bar(title=f'variable {tipo_var}')
            
        elif tipo_var == "numerica_continua":
            my_df[col].plot.box(title=f'variable {tipo_var}')

        else:
            serie_years = my_df[col].dt.year.value_counts().sort_values(ascending=False)
            if len(serie_years > 10):
                serie_years = serie_years[:10]
                serie_years.plot.bar(title=f'variable {tipo_var} (top 10)')
            else:
                serie_years.plot.bar(title=f'variable {tipo_var}')

        # Guardamos el grafico en archivo
        file_path = os.path.join(carpeta, 'columna_' + col + '.png')
        plt.tight_layout()
        plt.savefig(file_path)
        plt.close()
        print(f'Gráfico guardado en: {file_path}')

```

```python
exportarGraficos(df_niv3_titanic, 'Graficos')
```

### **Ejercicio 2.**

Carga el archivo matriz_distancies.xlsx en pandas, de modo que los nombres de filas y los nombres de columnas sean los de las ciudades. Borra "Las Palmas de Gran Canaria" y "Palma" para que podamos realizar el trayecto en coche.

Fuente: Mejoras Rutas

Nos interesa visitar todas las principales ciudades de España recorriendo la mínima distancia posible.

No hace falta que lo hagas de forma óptima, nos interesa que desarrolles una solución razonable utilizando las herramientas que tienes actualmente.

Por ejemplo, una aproximación sencilla (que no óptima) sería yendo siempre a la ciudad más cercana que no hayamos visitado todavía

Haz una función que dada la matriz de distancias y la ciudad de origen, haga una propuesta de ruta que sea lo más corta posible que puedas, devolviendo una lista con el orden de visita. Da también la distancia total recorrida.

```python
df_dist = pd.read_excel("matriu_distancies.xlsx")
df_dist
```

```python
df_dist = df_dist.rename(columns={'Unnamed: 0': 'Ciudad'})
df_dist = df_dist.set_index('Ciudad')

# Borramos la fila y la columna "Las Palmas de Gran Canaria"
df_dist.drop(columns=['Las Palmas de Gran Canaria', 'Palma'], index=['Las Palmas de Gran Canaria', 'Palma'], inplace=True)

# Cambiamos los nans por np.inf 
df_dist.fillna(np.inf, inplace=True)
```

```python
print(df_dist.shape)
df_dist
```

```python
def obtenerRutaMasCorta(df_dist, ciudad):

    df_dist_copy = df_dist.copy()

    # Asignamos inf a los valores de la columna "ciudad" para que el algoritmo no regrese a la primera ciudad
    df_dist_copy[ciudad] = np.inf 

    # Agregamos la primera ciudad en la lista de recorrido
    recorrido = [ciudad]
    dist_recorrido = 0
    desde_ciudad = ciudad

    while len(df_dist_copy) > 1:
        min_dist = np.inf
        fila_desde_ciudad = df_dist_copy.loc[desde_ciudad]

        # Buscamos la ciudad más cercana
        for i in range( len(fila_desde_ciudad) ):
            if fila_desde_ciudad[i] < min_dist:
                min_dist = fila_desde_ciudad[i]
                hasta_ciudad = fila_desde_ciudad.index[i]

        # Agregamos la ciudad más cercana a la lista de recorrido
        recorrido.append(hasta_ciudad)
        dist_recorrido += min_dist

        # Quitamos la fila "desde_ciudad" y la columna "hasta_ciudad" ya que este tramo esta agregado en el recorrido
        df_dist_copy.drop(index=desde_ciudad, inplace=True)
        df_dist_copy.drop(columns=hasta_ciudad, inplace=True)

        # Ya que la "hasta_ciudad" ya esta en el recorrido asignamos inf a su distancia hasta "desde_ciudad" 
        # para excluirla desde el calculo de minimo
        if desde_ciudad in df_dist_copy.columns:
            df_dist_copy.loc[hasta_ciudad, desde_ciudad] = np.inf

        desde_ciudad = hasta_ciudad

    return recorrido, dist_recorrido
```

```python
obtenerRutaMasCorta(df_dist, 'Barcelona')
```

(EXTRA) Desde qué ciudad la ruta sería más corta con el algoritmo planteado.

```python
def obtener_dist_más_corta_y_ciudad(): 

    lista_rutas = []
    dist_más_corta = np.inf
    
    for ciudad in df_dist:
        lista_rutas.append(obtenerRutaMasCorta(df_dist, ciudad))

    for fila in lista_rutas:
        if fila[1] < dist_más_corta:
            dist_más_corta = fila[1]
            desde_que_ciudad = fila[0][0]

    return dist_más_corta, desde_que_ciudad
```

```python
distancia_min, ciudad_min = obtener_dist_más_corta_y_ciudad()
print(f"Distancia más corta será {distancia_min} desde {ciudad_min}")
```
