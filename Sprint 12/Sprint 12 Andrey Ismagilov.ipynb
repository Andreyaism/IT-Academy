### **Ejercicio 4.**  
### Realiza una petición POST para crear una nueva publicación ficticia. Incluye el título, el cuerpo del mensaje y un userId.

##### Obtenemos el número de publicaciones del recurso POSTS

```python
r_posts = requests.get(url_posts)
lista_posts = r_posts.json()
num_de_public = len(lista_posts)
```

##### Creamos  la nueva publicación ficticia

```python
public_ficticia = {
    'userId': 1,
    'id': num_de_public + 1,
    'title': 'title fictitious AI',
    'body': 'message fictitious AI'
    }

r_de_post = requests.post(url_posts, data = public_ficticia)

if r_de_post.status_code == 201:
    print('Exito.')
    print(f'Codico de estado: {r_de_post.status_code}')
    print('Cuerpo de publicación: \n', r_de_post.request.body)
else:
    print(f'Error. Codico de error: {r_de_post.status_code}')
```

### **Ejercicio 5.**  
###  Realiza una petición PATCH para modificar parcialmente una publicación existente.

```python
id_post_a_modificar = 1

url_a_modificar = url_posts + '/' + str(id_post_a_modificar)

public_ficticia_modificada = {
    "body": "Modificada por PATCH"
  }

r_de_patch = requests.patch(url_a_modificar, data = public_ficticia_modificada)

if r_de_patch.status_code == 200:
  print('Modificado.')
  print(f'Codico de estado: {r_de_patch.status_code}')
else:
  print(f'Error. Codico de error: {r_de_patch.status_code}')
```

```python
df_flights = pd.json_normalize(df_vuelos.flight).add_prefix('flight_')
df_flights.head(3)
```

```python
df_aeronaves = pd.json_normalize(df_vuelos['aircraft']).add_prefix('aircraft')
df_aeronaves.head(3)
```

Unimos los datafames creados al unico.

```python
df_vuelos_total = pd.concat([df_salidas, df_llegadas, df_aerolinias, df_flights, df_aeronaves], axis=1)
df_vuelos_total.head(5)
```

#### Utiliza la API de Open Data BCN para Buscar un dataset de tu interés mediante package_search y package_show.

Miramos la lista dels datasets usando **package_search**.

```python
url_bcn = 'https://opendata-ajuntament.barcelona.cat/data/api/action/package_search'

params = {'q': '*:*', 'rows': 10}

r_openBCN = requests.get(url_bcn, params=params)

if r_openBCN.status_code == 200:
    resultados = r_openBCN.json()['result']['results']
    for ds in resultados:
        print(f"Título: {ds['title']} | ID: {ds['name']}")
```

Hacemos una búsqueda más concreta, por ejemplo los nombres más populares.

```python
params = {'q': 'Most popular names', 'rows': 10}

r_openBCN = requests.get(url_bcn, params=params)

if r_openBCN.status_code == 200:
    results = r_openBCN.json()['result']['results']
    for ds in results:
        print(f"Título: {ds['title']} | {ds['id']} |  ID: {ds['name']}")
```
