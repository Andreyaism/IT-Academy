#### 1. Calculadora del índice de masa corporal

- Escribe una función que calcule el IMC ingresado por el usuario/a, es decir, quien lo ejecute deberá ingresar estos datos. Puedes obtener más información de su cálculo en: 🔗 Índice de masa corporal IMC que es y cómo se calcula.

- La función debe clasificar el resultado en sus respectivas categorías.

Consejo: Intenta validar los datos previamente, para que envíe un mensaje de advertencia si los datos introducidos por el usuario/a no están en el formato adecuado o no toma valores razonables.

```python

def inputNumValidado(prompt_num):
    # Función para validar el parametro introducido
    # Recibe como el argumento "prompt_num" de tipo string.
    # Devuelve el número introducido en caso si la validacion sea exitosa o "-1" en caso contrario.
    try:
        num = float(input(prompt_num))
    except ValueError :
        print("Error: el parametro debe ser numero entero o decimal")
        return -1
    if num <= 0:
        print("Error: el parametro debe ser más de zero")
        return -1
    return num

def clasificarPeso(peso, altura):
    # Función para calcular IMC y clasificar el peso
    imc = round(peso/altura**2, 1)
    if imc < 18.5:
        rango = "peso bajo"
    elif imc < 25:
        rango = "peso normal"
    elif imc < 30:
        rango = "sobrepeso"
    else:
        rango = "obesidad"
    return imc, rango
```

```python
def main_imc():
    print("""    ***********************************************
    *** Calculadora del índice de masa corporal ***
    ***********************************************""")
    tu_peso = -1
    tu_altura = -1
    while tu_peso == -1:
        tu_peso = inputNumValidado("Tu peso en kilos:")
    while tu_altura == -1:
        tu_altura = inputNumValidado("Tu altura en metros:")
    tuIMC, tuRango = clasificarPeso(tu_peso, tu_altura)

    print(f"Tu peso: {tu_peso} kg, Tu altura: {tu_altura} metros")
    print(f"Tu IMC: {tuIMC}. Tienes {tuRango}")
```

2. Conversor de temperaturas

Existen diversas unidades de temperatura utilizadas en distintos contextos y regiones. Las más comunes son Celsius (°C), Fahrenheit (°F) y Kelvin (K). También existen otras unidades como Rankine (°Ra) y Réaumur (°Re).

Selecciona al menos 2 conversores, de modo que al introducir una temperatura devuelva, como mínimo, dos conversiones, de modo que se puedan guardar (recuerda que un print() no se puede guardar nunca).

Consejo: Intenta validar los datos previamente, para que envíe un mensaje de advertencia si los datos introducidos por el usuario/a no están en el formato adecuado.

(EXTRA): Piensa una manera de almacenar todas las posibles conversiones en un solo objeto (¿Lista? ¿Diccionario? ¿DataFrame?) en vez de escribir muchos if else en función de la temperatura de origen y la temperatura de destino.

```python
def validarTemperatura(prompt_temp):
    es_temperatura_string = True
    while es_temperatura_string:
        try:
            temperatura = float(input(prompt_temp))
            es_temperatura_string = False
            return temperatura
        except:
            print("Error: el parametro debe ser en numero entero o decimal")
    

def validarUnidad(prompt_unidad):
    es_unidad_no_correcta = True
    unidad_opcion = ["celsius", "fahrenheit", "kelvin", "rankine", "reaumur"]
    while  es_unidad_no_correcta:
        unidad = input(prompt_unidad)
        if unidad in unidad_opcion:
            es_unidad_no_correcta = False
        else:
            print("Error: unidad debe ser ", unidad_opcion)    
    return unidad


def conversor_a_celsius(t, u):
    A_CELSIUS = {
        "celsius": lambda t: t,
        "fahrenheit": lambda t: round((t - 32) / 1.8, 2),
        "kelvin": lambda t: t - 273.15,
        "rankine": lambda t: round((t - 491.67) / 1.8, 2),
        "reaumur": lambda t: round(t / 0.8, 2)
                }
    return A_CELSIUS[u](t)

def conversor_desde_celsius(t, u):
    DESDE_CELSIUS = {
        "celsius": lambda t: t,
        "fahrenheit": lambda t: t * 1.8 + 32,
        "kelvin": lambda t: t + 273.15,
        "rankine": lambda t: (t + 273.15) * 1.8,
        "reaumur": lambda t: t * 0.8 
        }
    return DESDE_CELSIUS[u](t)

def conversor_a_fahrenheit(t, u):
    A_FAHRENHEIT = {
        "celsius": lambda t: round(t*1.8 + 32, 2),
        "fahrenheit": lambda t: t,
        "kelvin": lambda t: round(t * 1.8 - 459.67,2),
        "rankine": lambda t: round(t - 459.67, 2),
        "reaumur": lambda t: round(t * 2.25 + 32, 2)
    }
    return A_FAHRENHEIT[u](t)

def conversor_desde_fahrenheit(t, u):
    DESDE_FAHRENHEIT = {
        "celsius": lambda t: round((t - 32)/1.8, 2),
        "fahrenheit": lambda t: t,
        "kelvin": lambda t: round((t + 459.67)/1.8, 2),
        "rankine": lambda t: round(t + 459.67, 2),
        "reaumur": lambda t: round((t - 32)/2.25, 2)
    }
    return DESDE_FAHRENHEIT[u](t)
```

```python
def executar_conversion():
    my_temperatura = validarTemperatura("Temperatura:")
    my_unidad = validarUnidad("Unidad de temperatura: celsius, fahrenheit, kelvin, rankine, reaumur")

    resultado1 = conversor_a_celsius(my_temperatura, my_unidad)
    print(f"{my_temperatura} {my_unidad} = {resultado1} celsius")

    resultado2 = conversor_desde_celsius(my_temperatura, my_unidad)
    print(f"{my_temperatura} celsius = {resultado2} {my_unidad}")

    resultado3 = conversor_a_fahrenheit(my_temperatura, my_unidad)
    print(f"{my_temperatura} {my_unidad} = {resultado3} fahrenheit")

    resultado4 = conversor_desde_fahrenheit(my_temperatura, my_unidad)
    print(f"{my_temperatura} fahrenheit = {resultado4} {my_unidad}")
```

Escribe una función que dado un texto, muestre las veces que aparece cada palabra. Intenta que se gestionen todas las posibles casuísticas que hagan que el programa no funcione correctamente.

(EXTRA): ¿Cuál es la longitud media de las palabras del texto que has escrito? “¿Hola cómo va?” debería devolver (4+3+2) / 3 = 3m

```python
def frase_a_diccionario(frase):

    lista = re.findall(r'\b\w+\b', frase.lower())

    diccionario = {}
    for palabra in lista:
        palabra_esta_en_diccionario = False
        # Comprobamos si la palabra de la lista hay en diccionario
        for k in diccionario.keys():
            if palabra == k:
                palabra_esta_en_diccionario = True
                break
        # Si no hay contamos cuantos veces se aparece en la lista     
        if palabra_esta_en_diccionario:
            continue
        else:
            counter = lista.count(palabra)
            diccionario[palabra] = counter
    diccionario_sorted = dict( sorted(diccionario.items(), key=lambda item: item[0]) )
    return(diccionario_sorted)

def obtenerMediaLongitudDePalabra(my_dict):
    if len(my_dict) == 0: 
        return 0
    else: 
        return round(st.mean(map(len, my_dict)), 0)
```

```python
def contador_de_palabras():
    # mi_frase = "En la casa de Pinocho todos cuentan hasta Ocho: Pin uno, piN dos, pin cuatro, pin cinco, pin seis, pin siete, pin ocho."
    mi_frase = input("Escribe una frase:")
    mi_dict = frase_a_diccionario(mi_frase)
    print(mi_dict)
    print(obtenerMediaLongitudDePalabra(mi_dict))
```

4. Diccionario inverso (con posibilidad de duplicados)

Resulta que el cliente tiene una encuesta muy antigua que se almacena en un diccionario y los resultados los necesita a la inversa, es decir, intercambiando las claves y valores. Los valores y claves en el diccionario original son únicos; si éste no es el caso, la función debería imprimir un mensaje de advertencia, junto con una lista con los valores asociados a la clave repetida.

Fíjate en los siguientes ejemplos para ver el output esperado.

```python
def reverse_dictionary(my_dict):
    new_dict = {}
    new_dict_tiene_valores_repetidos = False
    for k, v in my_dict.items():
        # Comprobamos si el diccionario tiene valores de tipo list, tuple, dict, set.
        if isinstance(v, (list, tuple, dict, set)):
            print("Error: no se puede crear el diccionario inverso. El diccionario recibido tiene valores de tipo colecciones de datos.")
            return
        new_dict_key = new_dict.get(v)
        if new_dict_key == None:
            new_dict[v] = k
        else:
            new_dict_tiene_valores_repetidos = True
            if isinstance(new_dict_key, list):
                new_dict[v].append(k)
            else:
                new_dict[v] = [new_dict_key, k]
    if new_dict_tiene_valores_repetidos == True: 
        print("Advertencia: diccionario tiene valores con claves repetidos")
    return(new_dict)
```

```python
def main_diccionarios_inversos():
    my_dict1 = {'a':1, 'b':2, 'c':3}
    print(my_dict1)
    my_dict_revers1 = reverse_dictionary(my_dict1)
    print(my_dict_revers1)
    print("***************************************************************************************************")
    my_dict2 = {'a':1, 'b':2, 'c':3, 'd':('s', 'g')}
    print(my_dict2)
    my_dict_revers2= reverse_dictionary(my_dict2)
    print(my_dict_revers2)
    print("***************************************************************************************************")
    my_dict3 = {'x':'apple', 'y':'banana', 'z':'banana', 'w': 'banana', 'q':'cereza'}
    print(my_dict3)
    my_dict_revers3 = reverse_dictionary(my_dict3)
    print(my_dict_revers3)
```

El cliente quedó contento con el contador de palabras, pero ahora quiere leer archivos TXT y que calcule la frecuencia de cada palabra ordenadas dentro de las entradas habituales del diccionario según la letra con la que comienzan, es decir, las claves deben ir de la A a la Z y dentro de la A debemos ir de la A a Z. "tu_me_quieres_blanca.txt" la salida esperada sería:

```python

def dict_a_dict_con_subdict(my_dict):
    # Función recibe como el argumento diccionario con palabras y número de veces que aparece la palabra
    # y devuelva el diccionario con subdiccionarios donde la clave de cada subdiccionario sea la letra alfabetica
    # con que comienza cada palabra del subdiccionario.
    primera_letra = list(my_dict.keys())[0][0]
    dict_with_subdict = { primera_letra: {} }
    for k, v in my_dict.items():
        if k[0] == primera_letra:
            dict_with_subdict[primera_letra][k] = v
        else:
            primera_letra = k[0]
            dict_with_subdict.update({k[0]: {k: v}})
    return(dict_with_subdict)

```

```python
def conversor_de_lista(my_list):
    # Función recibe como el argumento lista y devuelva tres listas:
    # la primera: la lista de números decimales
    # la segunda: la lista de otras elementos
    # la tercera: la lista de elementos de tipo dict y subelementos de tipo dict, list, set, tuple
    lista_de_numeros = []
    lista_de_otras = []
    lista_de_dict = []
    for elem in my_list:
        if isinstance(elem, (list, tuple, set)):
            nums, otras, dicts = conversor_de_lista(elem)
            lista_de_numeros.extend(nums)
            lista_de_otras.extend(otras)
            lista_de_dict.extend(dicts)

        elif isinstance(elem, dict):
            lista_de_dict.append(elem)
        
        else:
            try:
                lista_de_numeros.append(float(elem))
            except (ValueError, TypeError):
                lista_de_otras.append(elem)
       
    return lista_de_numeros, lista_de_otras,  lista_de_dict
```

1. Generador de Contraseñas

Explora el funcionamiento del módulo random de la librería numpy. 🔗 Random module in NumPy

En este punto, el cliente ha detectado un problema con las contraseñas que utilizan sus trabajadores. Asdf1234, fechas de cumpleaños o similares. Para solucionarlo, nos ha encargado una función de Python que genere contraseñas más seguras. La función debe depender de los siguientes parámetros:

- longitud (int): Longitud de la contraseña
- mayúsculas (bool = True): Si tiene que aparecer mayúsculas
- minúsculas (bool = True): Si tiene que aparecer minúsculas
- números (bool = True): Si tiene que aparecer números
- signos (bool = False): Si tiene que aparecer caracteres especiales (,-$? o similares)

Así pues, si ejecutamos la función de la siguiente forma:

crear_contraseña(10, True, True, True, True)

Deberíamos obtener un output (que debemos poder guardar) del estilo:

9Er,5Vn8P$

Asegúrate de que se cumplan todos los criterios, y que estas contraseñas sean realmente aleatorias.

(EXTRA) Explora cómo podríamos hacer que la función copiara la contraseña automáticamente en el portapapeles del ordenador (como si la hubiéramos seleccionado y hecho ctrl+copy).

```python
def crear_contraseña(longitud_contr, hay_mayusculas, hay_minuskulas, hay_numeros, hay_symb_especiales):
    symb_mayusculas = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
    symb_minusculas = 'abcdefghijklmnopqrstuvwxyz'
    numeros = '0123456789'
    symb_especiales = '!"#$%&*+,-./:;?@^_'

    symb_para_generar_contraseña = ""
    parte_obligatoria = []
    longitud_maxima = len(symb_mayusculas + symb_minusculas + numeros + symb_especiales)

    if longitud_contr < 4:
        print("Longitud minima de contraseña debe ser 4 characteres.")
        return
    if not (hay_mayusculas + hay_minuskulas + hay_numeros + hay_symb_especiales):
        print("No se puede generar contraseña.")
        return
    if longitud_contr > longitud_maxima:
        print(f"Longitud maxima de contraseña debe ser menos de {longitud_maxima}.")
        return
    
    if hay_mayusculas: 
        symb_para_generar_contraseña = symb_para_generar_contraseña + symb_mayusculas
        parte_obligatoria.append( symb_mayusculas[np.random.randint(len(symb_mayusculas))] )
    if hay_minuskulas: 
        symb_para_generar_contraseña = symb_para_generar_contraseña + symb_minusculas
        parte_obligatoria.append( symb_minusculas[np.random.randint(len(symb_minusculas))] )
    if hay_numeros:
        symb_para_generar_contraseña = symb_para_generar_contraseña + numeros
        parte_obligatoria.append( numeros[np.random.randint(len(numeros))] )
    if hay_symb_especiales:
        symb_para_generar_contraseña = symb_para_generar_contraseña + symb_especiales
        parte_obligatoria.append( symb_especiales[np.random.randint(len(symb_especiales))] )
    list_symb = list(symb_para_generar_contraseña)
    resto_de_contraseña_lista = sample(list_symb, longitud_contr - len(parte_obligatoria))
    contraseña = parte_obligatoria + resto_de_contraseña_lista
    contraseña_array = np.asarray(contraseña)
    random.shuffle(contraseña_array)
    return("".join(str(x) for x in contraseña_array))
```

```python
def main_crear_contraseña():
    contraseña = crear_contraseña(10, True, True, True, True)
    print("La contraseña generada: ", contraseña)
    print("Se la ha copiada en el portapapeles")
    pyperclip.copy(contraseña)
    print(pyperclip.paste())
    print("y la pegada luego.")
```

He escrito una función para automatizar el procesamiento de datos del fichero.

```python
def clasificadorDeEquipos(fichero):
    df = pd.read_csv(fichero, sep='\t', header=None)

    df['gols_marcados_1'] = df[1].str[0]
    df['gols_encajados_1'] = df[1].str[2]
    df['gols_marcados_2'] = df[1].str[2]
    df['gols_encajados_2'] = df[1].str[0]
    df['puntos_1'] = df.apply( lambda x: getPuntos( x['gols_marcados_1'], x['gols_encajados_1'] ), axis=1 )
    df['puntos_2'] = df.apply( lambda x: getPuntos( x['gols_marcados_2'], x['gols_encajados_2'] ), axis=1 )

    df['partido_ganado_1'] = 0
    df['partido_empatado_1'] = 0
    df['partido_perdido_1'] = 0
    df.loc[ df['gols_marcados_1'] > df['gols_encajados_1'],  'partido_ganado_1'] = 1
    df.loc[ df['gols_marcados_1'] == df['gols_encajados_1'],  'partido_empatado_1'] = 1 
    df.loc[ df['gols_marcados_1'] < df['gols_encajados_1'],  'partido_perdido_1'] = 1

    df['partido_ganado_2'] = 0 
    df['partido_empatado_2'] = 0
    df['partido_perdido_2'] = 0
    df.loc[ df['gols_marcados_2'] > df['gols_encajados_2'],  'partido_ganado_2'] = 1
    df.loc[ df['gols_marcados_2'] == df['gols_encajados_2'],  'partido_empatado_2'] = 1 
    df.loc[ df['gols_marcados_2'] < df['gols_encajados_2'],  'partido_perdido_2'] = 1


    df = df.rename(columns={ 0: 'equipo_1', 2:'equipo_2' })


    # Cambiamos los tipos de columnas
    df['gols_marcados_1'] = df['gols_marcados_1'].astype('int')
    df['gols_encajados_1'] = df['gols_encajados_1'].astype('int')
    df['gols_marcados_2'] = df['gols_marcados_2'].astype('int')
    df['gols_encajados_2'] = df['gols_encajados_2'].astype('int')
    df['puntos_1'] = df['puntos_1'].astype('int')
    df['puntos_2'] = df['puntos_2'].astype('int')
    df['equipo_1'] = df['equipo_1'].astype("string")
    df['equipo_2'] = df['equipo_2'].astype("string")

    df_equipo1 = df[['equipo_1', 'gols_marcados_1', 'gols_encajados_1', 'puntos_1', 'partido_ganado_1', 'partido_empatado_1', 'partido_perdido_1']]
    df_equipo2 = df[['equipo_2', 'gols_marcados_2', 'gols_encajados_2', 'puntos_2', 'partido_ganado_2', 'partido_empatado_2', 'partido_perdido_2']]

    df_equipo1 = df_equipo1.rename( columns = {'equipo_1': 'equipo', 'gols_marcados_1':'gols_marcados', \
                                                        'gols_encajados_1': 'gols_encajados', 'puntos_1':'puntos', \
                                                            'partido_ganado_1': 'partido_ganado', 'partido_empatado_1': 'partido_empatado', \
                                                                'partido_perdido_1': 'partido_perdido'} )

    df_equipo2 = df_equipo2.rename( columns = {'equipo_2': 'equipo', 'gols_marcados_2':'gols_marcados', \
                                                        'gols_encajados_2': 'gols_encajados', 'puntos_2':'puntos', \
                                                            'partido_ganado_2': 'partido_ganado', 'partido_empatado_2': 'partido_empatado', \
                                                                'partido_perdido_2': 'partido_perdido'} )
    
    df_result = pd.concat([df_equipo1, df_equipo2], ignore_index=True)
    clasificacion = df_result.groupby(['equipo']).sum().sort_values(by="puntos", ascending=False)
    return clasificacion
```

```python
def main_clasificador_de_equipos():
    file_path = 'historic partits.txt'
    print(clasificadorDeEquipos(file_path))
```
